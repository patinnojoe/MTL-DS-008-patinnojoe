```python
from src.config import PROCESSED_DATA_PATH, ANALYTICAL_DATA_PATH
import pandas as pd


processed_files = PROCESSED_DATA_PATH.glob("incomplete_provider_*.csv")

dfs = [pd.read_csv(file, parse_dates=['period'])  for file in processed_files]

# combine all processed 
df = pd.concat(dfs, ignore_index=True)





```

```python


# create readable date time labels
# set year, month, year and month labels

df["month_name"] = df["period"].dt.strftime("%b")
df['month'] = df['period'].dt.month
df['year'] = df['period'].dt.year
df["month_label"] = df["period"].dt.strftime("%b %Y")
```

```python
# mask = ( df['Treatment Function Code'] == 'C_110') & (df['Provider Code'] == 'RXN' ) & (df['year'] == 2023)
# df[mask]
# april 2023 uses xls rather than xlsx extensions that other files used, i had to install xlrd in addition to openxly to read that excel format
april_2023 = df[
    (df["year"] == 2023) &
    (df["period"].dt.month == 4)
]

print("Total records:", len(april_2023))
```
